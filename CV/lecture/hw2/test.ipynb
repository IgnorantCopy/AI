```python
import cv2
```

```python
def show_img(img):
    cv2.imshow('img', img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
img = cv2.cvtColor(cv2.imread('data/img07.jpg'), cv2.COLOR_BGR2GRAY)
show_img(img)
```

```python
edges = cv2.Canny(cv2.GaussianBlur(img, ksize=(0, 0), sigmaX=1), 40, 150)
show_img(edges)
```

```python
img = cv2.cvtColor(cv2.imread('data/img01.jpg'), cv2.COLOR_BGR2GRAY)
show_img(img)
```

```python
blur = cv2.GaussianBlur(img, (0, 0), 3)
Ix = cv2.Sobel(blur, cv2.CV_64F, 1, 0, ksize=3)
Iy = cv2.Sobel(blur, cv2.CV_64F, 0, 1, ksize=3)
```

```python
show_img(Ix)
```

```python
show_img(Iy)
```
