# torch.nn

## nn.Module

常用方法：
+ add_module(name, module): 添加子模块
+ apply(fn): 应用函数到所有子模块
+ bfloat16(): 转换为浮点类型参数为 bfloat16 类型
+ buffers(recurse=True): 返回所有缓冲的迭代器，如果 recurse=True，则递归返回所有子模块的缓冲区 （buffer 是不需要梯度的 tensor，区别于 parameter）
+ children(): 返回所有直接子模块的迭代器
+ cpu(): 将所有参数和缓冲区移动到 CPU 上
+ cuda(device=None): 将所有参数和缓冲区移动到 GPU 上
+ get_parameter(target): 获取参数
+ get_submodule(target): 获取子模块
+ named_parameters(prefix='', recurse=True): 返回所有参数及其名称的迭代器
+ parameters(recurse=True): 返回所有参数的迭代器
+ requires_grad_(requires_grad=True): 设置参数的 requires_grad 属性
+ state_dict(): 返回模型状态字典
